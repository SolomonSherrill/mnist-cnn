```python
import torch
import torch.nn.functional as F
from torchvision import datasets
from torchvision.transforms import v2

ds = datasets.FashionMNIST(
    root="/Users/solomon/mnist-cnn/data",
    train=True,
    download=True,

    #turns PIL image into a torchvision image tensor, and casts 8 bit ints into floats 0-1, representing intensity of each pixel
    transform=v2.Compose([v2.ToImage(), v2.ToDtype(torch.float32, scale=True)]),
    #turns int labels 1-9 into a onehot tensor with 10 classes, and casts to a float as it is the expected datatype
    target_transform=v2.Lambda(
        lambda y: F.one_hot(torch.tensor(y), num_classes=10).float()
    ),
)
```
